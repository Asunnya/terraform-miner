```python
import pandas as pd

url = 'https://raw.githubusercontent.com/Asunnya/terraform-miner/refs/heads/master/src/nlp_analysis/analysis_results_nlp/commits_com_clusters.jsonl'
df = pd.read_json(url, lines=True)
print(len(df))
display(df.head())
```

```python
df[['owner', 'name', 'gibberish']] = df['repo_name'].str.split('_', n=2, expand=True)
df = df.drop(columns=['gibberish'])
display(df.head())
```

```python
df['repo_URL'] = 'https://github.com/' + df['owner'] + '/' + df['name']
df['commit_URL'] = "=HYPERLINK(\""+df['repo_URL']+'/commit/'+df['commit_hash']+"\")"
display(df.head())
```

```python
df_sampled = df.sample(n=169, random_state=42)
display(df_sampled.head())
display(df_sampled.shape)
```

```python
import numpy as np
# sample size: used Cochran with 99% confidence level, 10% margin of error = 167.
n_total = 169
n_unique_per_file = 38
n_overlapping = 17
n_per_file = n_unique_per_file + n_overlapping
df_sampled['is_bugfix'] = ""
df_sampled['obs'] = ""
df_sampled = df_sampled[['commit_URL','is_bugfix', 'obs']]
all_indices = df_sampled.index.tolist()

np.random.shuffle(all_indices)

overlapping_indices = all_indices[:n_overlapping]

unique_indices = all_indices[n_overlapping:]

unique_indices_split = np.array_split(unique_indices, 4)

indices_file1 = overlapping_indices + unique_indices_split[0].tolist()
indices_file2 = overlapping_indices + unique_indices_split[1].tolist()
indices_file3 = overlapping_indices + unique_indices_split[2].tolist()
indices_file4 = overlapping_indices + unique_indices_split[3].tolist()

df_file1 = df_sampled.loc[indices_file1]
df_file2 = df_sampled.loc[indices_file2]
df_file3 = df_sampled.loc[indices_file3]
df_file4 = df_sampled.loc[indices_file4]

df_file1.to_csv('sample_heleno.csv', index=False)
df_file2.to_csv('sample_isadora.csv', index=False)
df_file3.to_csv('sample_vania.csv', index=False)
df_file4.to_csv('sample_joao.csv', index=False)
```
